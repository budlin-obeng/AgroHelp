# Space2Soil: Phase 3 Scientific Analysis & Agricultural Indicators
### Agroclimatological Models, Thermal Regimes & Crop Suitability for Ejura, Ghana
**Project:** Space2Soil (NASA Space Apps Challenge: "Field Shift: Adapting Farms with NASA Data")  
**Location:** Ejura, Ashanti Region, Ghana (`7.3856° N, -1.3562° W`)  
**Decadal Scope:** 2016 – 2025 (10 Complete Years)

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Setup plot styles
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 11

analysis_dir = os.path.abspath(os.path.join('..', 'data', 'processed', 'analysis'))
print('Loading Phase 3 analytical outputs from:', analysis_dir)
```

## 1. Agricultural Seasonality & Growing Period Trends

```python
df_season = pd.read_csv(os.path.join(analysis_dir, 'rainfall_seasonality_metrics.csv'))

fig, ax1 = plt.subplots(figsize=(11, 5))
ax1.plot(df_season['year'], df_season['major_season_onset_doy'], marker='o', color='#2ca02c', linewidth=2.5, label='Major Season Onset (Day of Year)')
ax1.set_xlabel('Year', fontweight='bold')
ax1.set_ylabel('Onset Day of Year (DOY)', color='#2ca02c', fontweight='bold')
ax1.tick_params(axis='y', labelcolor='#2ca02c')

ax2 = ax1.twinx()
ax2.bar(df_season['year'], df_season['length_of_growing_period_days'], alpha=0.3, color='#1f77b4', width=0.4, label='Length of Growing Period (Days)')
ax2.set_ylabel('Total LGP (Days)', color='#1f77b4', fontweight='bold')
ax2.tick_params(axis='y', labelcolor='#1f77b4')
ax2.grid(False)

plt.title('Decadal Season Onset Date & Length of Growing Period (Ejura, Ghana)', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

df_season[['year', 'major_season_onset_date', 'season_cessation_date', 'length_of_growing_period_days', 'total_annual_rainfall_mm']]
```

## 2. Dry Spell Frequency & Severity Distribution

```python
df_spells = pd.read_csv(os.path.join(analysis_dir, 'dry_spell_events.csv'))

plt.figure(figsize=(10, 5))
severity_counts = df_spells['severity'].value_counts()
colors = {'Mild_Dry_Spell': '#a1d99b', 'Moderate_Moisture_Stress': '#fdae6b', 'Severe_Drought_Risk': '#e6550d'}
bar_colors = [colors.get(s, '#3182bd') for s in severity_counts.index]

plt.bar(severity_counts.index.str.replace('_', ' '), severity_counts.values, color=bar_colors, edgecolor='black', width=0.5)
plt.title('Distribution of Growing Season Dry Spells (>= 5 Days) [2016-2025]', fontsize=13, fontweight='bold')
plt.ylabel('Event Count', fontweight='bold')
plt.xlabel('Dry Spell Classification', fontweight='bold')
plt.tight_layout()
plt.show()

print(f'Total in-season dry spells identified: {len(df_spells)}')
df_spells.head(8)
```

## 3. Thermal Warming & Heat Stress Days

```python
df_thermal = pd.read_csv(os.path.join(analysis_dir, 'annual_temperature_trends.csv'))

fig, ax = plt.subplots(figsize=(11, 5))
ax.bar(df_thermal['year'] - 0.15, df_thermal['heat_stress_days_ge_35c'], width=0.3, color='#d95f02', label='Heat Stress Days (Tmax >= 35°C)')
ax.bar(df_thermal['year'] + 0.15, df_thermal['severe_heat_stress_days_ge_38c'], width=0.3, color='#7570b3', label='Severe Shock Days (Tmax >= 38°C)')
ax.set_xlabel('Year', fontweight='bold')
ax.set_ylabel('Number of Days', fontweight='bold')
ax.set_title('Annual Heat Stress Days Critical for Maize Pollination & Flowering', fontsize=13, fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()

df_thermal[['year', 'temp_mean_annual_c', 'temp_max_absolute_c', 'heat_stress_days_ge_35c', 'decadal_warming_rate_tmean_c_per_decade']]
```

## 4. Multi-Criteria Crop Suitability Scoring

```python
df_matrix = pd.read_csv(os.path.join(analysis_dir, 'crop_suitability_matrix.csv'))

plt.figure(figsize=(10, 5))
bars = plt.barh(df_matrix['crop'], df_matrix['mean_score'], color='#2ca02c', edgecolor='black', height=0.55)

# Add score labels
for bar in bars:
    w = bar.get_width()
    plt.text(w - 6, bar.get_y() + bar.get_height()/2, f'{w:.1f}', ha='center', va='center', color='white', fontweight='bold')

plt.axvline(80, color='darkgreen', linestyle='--', label='S1 Threshold (Highly Suitable)')
plt.axvline(65, color='orange', linestyle='--', label='S2 Threshold (Moderately Suitable)')
plt.xlabel('Composite Agro-climatic Suitability Score (0 - 100)', fontweight='bold')
plt.title('Crop Suitability Rankings for Ejura, Ashanti Region (Decadal Average)', fontsize=13, fontweight='bold')
plt.xlim(40, 105)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()

df_matrix
```
